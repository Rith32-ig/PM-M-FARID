**C.Latihan Langkah**

```python
import pandas as pd
data = {'luas':[36,70,45,90,54], 'kamar':[2,3,2,4,3],
        'lokasi':['Pinggir','Tengah','Pinggir','Tengah','Pinggir'],
        'harga':[250,600,320,800,400]}  # LABEL
df = pd.DataFrame(data)
X = df[['luas','kamar','lokasi']]   # fitur
y = df['harga']                     # label
```

```python
# Label numerik kontinu -> REGRESI
print(df['harga'].tolist())
# Label kategori -> KLASIFIKASI
df['kategori'] = ['Murah','Mahal','Murah','Mahal','Murah']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
    X, y, test_size=0.4, random_state=42)
print('Latih:', len(X_train), ' Uji:', len(X_test))
```

1. Supervised learning menggunakan data yang sudah memiliki label atau jawaban. Contohnya, memprediksi apakah siswa lulus atau tidak lulus berdasarkan jam belajar.
Unsupervised learning menggunakan data tanpa label dan bertujuan untuk menemukan pola atau kelompok yang ada dalam data.
Reinforcement learning belajar berdasarkan tindakan yang dilakukan dan hasil yang diperoleh, sehingga sistem dapat menentukan tindakan yang lebih baik.
