**C.Latihan Langkah**

```python
import pandas as pd
data = {'luas':[36,70,45,90,54], 'kamar':[2,3,2,4,3],
        'lokasi':['Pinggir','Tengah','Pinggir','Tengah','Pinggir'],
        'harga':[250,600,320,800,400]}  # LABEL
df = pd.DataFrame(data)
X = df[['luas','kamar','lokasi']]   # fitur
y = df['harga']                     # label
```

```python
# Label numerik kontinu -> REGRESI
print(df['harga'].tolist())
# Label kategori -> KLASIFIKASI
df['kategori'] = ['Murah','Mahal','Murah','Mahal','Murah']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
    X, y, test_size=0.4, random_state=42)
print('Latih:', len(X_train), ' Uji:', len(X_test))
```

**E.REFLEKSI**

1. Supervised learning adalah pembelajaran menggunakan data yang sudah memiliki label, sehingga model belajar dari data tersebut untuk menghasilkan prediksi. Contohnya memprediksi apakah seorang siswa lulus atau tidak.
Unsupervised learning adalah pembelajaran menggunakan data yang tidak memiliki label. Model akan mencari pola atau kelompok yang terdapat di dalam data tersebut.
Reinforcement learning adalah pembelajaran dengan cara memberikan reward (hadiah) atau konsekuensi dari tindakan yang dilakukan. Dari hasil tersebut, sistem belajar untuk menentukan tindakan yang lebih baik.

2. Pembagian data latih dan uji penting untuk mengetahui apakah model mampu bekerja dengan data yang belum pernah dilihat sebelumnya. Data latih digunakan untuk proses pembelajaran, sedangkan data uji digunakan untuk menguji hasil dan kemampuan generalisasi model.

3. Alur kerja ML disebut iteratif karena prosesnya dilakukan secara berulang. Setelah data diproses dan model diuji, hasilnya dapat dievaluasi. Jika hasilnya belum sesuai, proses dapat diperbaiki dan dijalankan kembali sampai mendapatkan hasil yang lebih baik.
